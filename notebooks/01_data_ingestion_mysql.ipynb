```python
pd.read_sql("""
SELECT activity_type, AVG(efficiency) AS avg_eff
FROM activities
GROUP BY activity_type
""", conn)
```

```python
pd.read_sql("""
SELECT day_of_week, AVG(burnout_risk) AS burnout
FROM activities
GROUP BY day_of_week
ORDER BY burnout DESC
""", conn)
```
